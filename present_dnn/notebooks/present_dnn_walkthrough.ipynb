# PRESENT-DNN Walkthrough

This notebook demonstrates PRESENT-80 encryption implemented as fixed-weight DNN operations.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
```

```python
from present_dnn.cipher.present_dnn import present_encrypt
from present_dnn.cipher.key_schedule import compute_all_round_keys
from present_dnn.utils.bit_utils import int_to_bitvec, bitvec_to_int
```

```python
pt = 0x0000000000000000
key = 0x00000000000000000000
ct = present_encrypt(pt, key, key_bits=80)
print(f'PT={pt:016x}')
print(f'KEY={key:020x}')
print(f'CT={ct:016x}')
```

```python
keys = compute_all_round_keys(int_to_bitvec(key, width=80))
print('K1 =', f'{bitvec_to_int(keys[0]):016x}')
print('K2 =', f'{bitvec_to_int(keys[1]):016x}')
print('K32=', f'{bitvec_to_int(keys[31]):016x}')
```
